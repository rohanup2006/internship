```python
import pandas as pd
df = pd.DataFrame({
 'Name': ['Abc', 'Bcd', 'Pqr', 'Xyz'],
 'Age': [32, 28, 45, 38],
 'Gender': ['Male', 'Female', 'Male', 'Female'],
 'Rating': [3.45, 4.6, 3.9, 2.78]},
 index=['r1', 'r2', 'r3', 'r4'])
print(df)
# Modify the Row labels of the DataFrame
df.index = [100, 200, 300, 400]
print(df)
```

```python
import pandas as pd
a = pd.DataFrame({'Name': ['Abc', 'Bcd', 'Pqr', 'Xyz'],
      'Age': [32, 28, 45, 38],
      'Gender': ['Male', 'Female', 'Male', 'Female'],
      'Rating': [3.45, 4.6, 3.9, 2.78]},
      index=['r1', 'r2', 'r3', 'r4'])
df = pd.DataFrame(a)
print(df)
df.columns = ['col1', 'col2', 'col3', 'col4']
print(df)
```
