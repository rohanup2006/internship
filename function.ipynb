```python
def maximum(a,b,c):
    return max(a,b,c)
maximum(10,29,30)
    
```

```python
def average(a,b,c,d):
    sum = a+b+c+d
    avg = sum / 4
    
    return avg
average(2,3,4,5)
```

```python
nums = [10,20,30,40,50]
def print_len(list):
    print(len(list))
print_len(nums)    

    
```

```python
nums = [10,20,30,40,50]
def print_list(list):
    for item in list:
        print(item,end=" ")
print_list(nums)
       
```

```python
def print_prod(list):
    result = 1
    for i in list:
        result = result * i
    return result
print_prod([10,20,30,40])
```

```python
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact *= i
    print(fact)
factorial(6)     
```

```python
def usd_to_rupee(usd_val):
    inr_val = usd_val*83
    print(usd_val,"usd=",inr_val,"inr")
usd_to_rupee(100)    
```
